**Objetivo:** Entender que los `enum` permiten definir un tipo de datos propio que solo puede tomar unos valores predeterminados.
**Fecha:** 2026-06-9

Los `enum` permiten usar variables definidas como Moneda.EURO o Moneda.DOLAR. 

```python
from enum import Enum

# Ejemplo 1: Enumeración Moneda
class Moneda(Enum):
    DOLAR = 1
    EURO = 2

def ejemplo_moneda():
    c = Moneda.DOLAR
    print(c)                # Moneda.DOLAR
    print(Moneda.DOLAR)     # Moneda.DOLAR
    print(type(c))          # <enum 'Moneda'>
    print("-" * 30)

    # Recorrer todos los miembros
    for c in Moneda:
        print(c)
    print("-" * 30)

# Ejemplo 2: Enumeración DiaSemana
class DiaSemana(Enum):
    LUNES = 1
    MARTES = 2
    MIERCOLES = 3
    JUEVES = 4
    VIERNES = 5
    SABADO = 6
    DOMINGO = 7

def ejemplo_dias():
    hoy = DiaSemana.MARTES
    print(hoy)        # DiaSemana.MARTES
    print(hoy.name)   # MARTES
    print(hoy.value)  # 2

    for dia in DiaSemana:
        print(dia.name, "=", dia.value)

    entrada = "VIERNES"
    if entrada in DiaSemana.__members__:
        print("Día válido:", entrada)
    else:
        print("Día inválido")
    print("-" * 30)

# Ejemplo 3: Menú interactivo con Moneda
class MonedaMenu(Enum):
    DOLAR = 1
    EURO = 2
    PESO = 3

def mostrar_menu():
    print("=== Selecciona una moneda ===")
    for c in MonedaMenu:
        print(f"{c.value}. {c.name}")

def ejemplo_menu():
    mostrar_menu()
    entrada = input("Ingresa el número de la moneda: ")

    if entrada.isdigit():
        opcion = int(entrada)
        try:
            moneda_seleccionada = MonedaMenu(opcion)
            print("Has elegido:", moneda_seleccionada.name)
        except ValueError:
            print("Número inválido, no corresponde a ninguna moneda.")
    else:
        print("Entrada inválida, debes ingresar un número.")

# Función principal que llama a todos los ejemplos
def main():
    ejemplo_moneda()
    ejemplo_dias()
    ejemplo_menu()

if __name__ == "__main__":
    main()
```
