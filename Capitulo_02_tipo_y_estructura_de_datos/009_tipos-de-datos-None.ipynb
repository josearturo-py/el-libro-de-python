Los `enum` son similares a null, nil u Option en otros lenguajes. 

```python
# Ejemplo práctico del uso de None en Python

def ejemplos_None():
    # None es un valor especial en Python que representa "ausencia de valor"
    x = None
    print(x)        # Muestra: None
    print(type(x))  # Muestra: <class 'NoneType'>
    print("-"*30)

    # None es un Singleton: siempre es el mismo objeto en memoria
    a = None
    b = None

    print(a is None)    # True → 'a' es None
    print(b is None)    # True → 'b' es None
    print(a is b)       # True → ambos apuntan al mismo objeto
    print(a == b)       # True → son iguales en valor
    print("-"*30)

    # Ejemplo de función que no devuelve nada explícitamente → retorna None
    def funcion():
        pass

    print(funcion())    # None
    print("-"*30)

    # Diccionarios: el método get() devuelve None si la clave no existe
    dic = {}
    print(dic.get('no existe'))   # None


def ejemplo_practico_None():
    # Ejemplo práctico: función que busca un usuario en una lista
    usuarios = ["Ana", "Luis", "Pedro"]

    def buscar_usuario(nombre):
        if nombre in usuarios:
            return nombre
        else:
            return None  # Si no lo encuentra, devuelve None

    # Caso encontrado
    resultado1 = buscar_usuario("Luis")
    print(f"Resultado de buscar 'Luis': {resultado1}")  # Luis

    # Caso no encontrado
    resultado2 = buscar_usuario("Carlos")
    if resultado2 is None:
        print("Usuario 'Carlos' no encontrado")  # Mensaje claro
    print("-"*30)


def main():
    ejemplos_None()
    ejemplo_practico_None()


if __name__ == "__main__":
    main()
```
